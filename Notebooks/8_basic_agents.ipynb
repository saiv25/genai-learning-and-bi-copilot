```python
from dotenv import load_dotenv
import os
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers(a: int, b: int) -> int:
    """Add two numbers together."""
    return a + b

@tool
def multiply_numbers(a: int, b: int) -> int:
    """Multiply two numbers together."""
    return a * b
```

```python
add_numbers.invoke({"a": 3, "b": 5})
```

```python
from langchain_groq import ChatGroq
from langchain.agents import create_agent
llm = ChatGroq(
    model="openai/gpt-oss-120b",
)
```

```python
agent = create_agent(
    model=llm,
    tools=[add_numbers,multiply_numbers], 
    system_prompt="You are a helpful assistant that can perform basic arithmetic operations like addition and multiplication."
)
```

```python
response = agent.invoke({"messages": [{"role": "user", "content": "What is 3 + 30 * 10?"}]})
```
