```python
from langchain_groq import ChatGroq
llm = ChatGroq(model="openai/gpt-oss-120b",)
text = "Hello, My name is Sainath Vangujare,"\
"Email is saivangujare@gmail.com and my age is 23 and i Live in Chhatrapati Sambhajinagar, Maharashtra, India. My phone number is 1234567890 and my pincode is 431001."
from pydantic import BaseModel, Field
class Responsestructured(BaseModel):
    name: str = Field(description="The name of the person")
    age: int = Field(description="The age of the person")
    city: str = Field(description="The city where the person lives")
    mail: str = Field(description="The email address of the person")
structured_llm =llm.with_structured_output(Responsestructured)
res = structured_llm.invoke("Please provide the name, age, city, and email address of the person in a structured format.")
res
```

```python
class Movies(BaseModel):
    title: str = Field(description="The title of the movie")
    year: int = Field(description="The release year of the movie")
    genre: str = Field(description="The genre of the movie")
    director: str = Field(description="The director of the movie")
movie_llm = llm.with_structured_output(Movies)
result = movie_llm.invoke("Give me one most watched thriller movie. Return the title, year, genre, and director in a structured format.")
print(result)
```
